```python
# Project Euler - Problem 1: Multiples of 3 or 5

def sum_divisible_by(n, limit):
    p = (limit - 1) // n
    return n * p * (p + 1) // 2

# O(1) Closed-form mathematical solution
LIMIT = 1000
total_math = sum_divisible_by(3, LIMIT) + sum_divisible_by(5, LIMIT) - sum_divisible_by(15, LIMIT)

print(f"Mathematical O(1) Solution: {total_math}")
```
